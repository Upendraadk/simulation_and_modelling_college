```python
import numpy as np
from scipy.stats import chi2

n = 100
k = 10

random_numbers = np.random.random(n)

observed, bins = np.histogram(
    random_numbers,
    bins=k,
    range=(0, 1)
)

expected = n / k

chi_square = np.sum((observed - expected) ** 2 / expected)

df = k - 1

alpha = 0.05

critical_value = chi2.ppf(1 - alpha, df)

print("\nObserved Frequencies:")
print(observed)

print("\nExpected Frequencies:")
print(expected)

print("\nChi-Square Calculated Value:")
print(chi_square)

print("\nDegree of Freedom:")
print(df)

print("\nCritical Value at 5% Significance Level:")
print(critical_value)

if chi_square < critical_value:
    print("\nResult: Fail to Reject H0")
    print("The generated random numbers appear to be uniformly distributed.")
else:
    print("\nResult: Reject H0")
    print("The generated random numbers do not appear to be uniformly distributed.")

```
