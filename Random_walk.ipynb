```python
import random
import matplotlib.pyplot as plt
steps=10
x=0
y=0
x_pos = [x]
y_pos = [y]
forward= 0
left=0
right=0
for _ in range(steps):
    r = random.uniform(0,10)
    if r < 5:
        y += 1
        forward += 1
    elif x < 7:
        x -= 1
        left += 1
    else:
        x += 1
        right += 1
    x_pos.append(x)
    y_pos.append(y)
print("Total Steps:", steps)
print("Forward:", forward)
print("Left:", left)
print("Right:", right)
plt.figure(figsize=(7,7))
plt.plot(x_pos, y_pos,marker='o', markersize=3)
plt.scatter(x_pos[0],y_pos[0],color='green',s=80,label='Start')
plt.scatter(x_pos[-1],y_pos[-1],color='red',s=80,label='End')
plt.title("Random Walk of a Drunk Person")
plt.xlabel("X Position")
plt.ylabel("Y Position")
plt.grid(True)
plt.legend()
plt.show()

plt.figure(figsize=(6,4))
plt.bar(["Forward","Left","Right"], [forward,left,right])
plt.title("Movement Distribution")
plt.ylabel("Number of moves")
plt.xlabel("Direction")
plt.grid(axis='y')
plt.show()
```
