```python
import numpy as np
import matplotlib.pyplot as plt

R = 1000
C = 0.001
Vo = 10
to = 0
T = 10
dt = 0.01
t = np.arange(to , T + dt , dt)
v = np.zeros(len(t))
v[0] = Vo
for i in range(len(t)-1):
    dv = -v[i] / (R * C)
    v[i+1] = v[i] + dv * dt
plt.figure(figsize=(8,5))
plt.plot(t,v, color="red", label="capacitor discharging")
plt.xlabel("Time")
plt.ylabel("Voltage")
plt.title("Capacitor Discharging Simulation")
plt.grid(True)
plt.legend()
plt.show()
```
