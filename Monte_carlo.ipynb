```python
import numpy as np
import matplotlib.pyplot as plt

N = 1000000

X = np.random.rand(N)
Y = np.random.rand(N)

inside = X**2 + Y**2 <= 1

Pi = 4 * np.sum(inside) / N

print("Estimated value of Pi:", Pi)
print("Actual value of Pi:", np.pi)

plt.figure(figsize=(6, 4))
plt.scatter(X[inside],Y[inside],marker='o',s=3,label='Inside circle')
plt.scatter(X[~inside],Y[~inside],marker='x',s=5,label='Outside circle')
plt.xlabel("X")
plt.ylabel("Y")
plt.title(f"Monte Carlo Simulation Estimation of Pi\n Estimated Pi = {Pi:.4f}")
plt.legend()
plt.axis("equal")
plt.grid(True)
plt.show()
```
